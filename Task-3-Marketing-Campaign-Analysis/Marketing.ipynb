```python
df.groupby("Location_ID")["Sales_Amount"].sum().sort_values(ascending=False).head(10)
```

```python
df["Store_Age_Group"] = pd.cut(
    df["Store_Age"],
    bins=[0,5,10,15,100],
    labels=["0-5 Years","6-10 Years","11-15 Years","15+ Years"]
)
```

```python
df["Sales_Category"] = pd.cut(
    df["Sales_Amount"],
    bins=[0,30,45,100],
    labels=["Low","Medium","High"]
)
```

```python
df.groupby("Promotion")["Sales_Amount"].sum().plot(kind="bar")
plt.title("Sales by Promotion")
plt.show()
```

```python
df.groupby("week")["Sales_Amount"].sum().plot(kind="line")
plt.title("Weekly Sales")
plt.show()
```

```python
df["Store_Age"].plot(kind="hist")
plt.title("Store Age Distribution")
plt.show()
```
