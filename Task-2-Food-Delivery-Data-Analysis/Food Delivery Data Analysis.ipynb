```python
df['Order_Status'].value_counts().plot(kind='bar')

plt.title('Orders by Status')
plt.xlabel('Order Status')
plt.ylabel('Number of Orders')
plt.xticks(rotation=0)
plt.show()
```

```python
df.groupby('Item_Name')['Total_Price'].sum().sort_values(ascending=False)
```

```python
df.groupby('Item_Name')['Total_Price'].sum().sort_values(ascending=False).plot(kind='bar')

plt.title('Total Order Value by Food Item')
plt.xlabel('Food Item')
plt.ylabel('Total Order Value')
plt.xticks(rotation=45)
plt.show()
```
